```python
import matplotlib.pyplot as plt
import numpy as np
from scipy.spatial import Delaunay
import math

# ----------------------------
# Data
# ----------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

points = np.array([p for seg in lines.values() for p in seg])

# ----------------------------
# Plot setup
# ----------------------------
fig, ax = plt.subplots(figsize=(6, 12))
ax.set_xlim(0, 20)
ax.set_ylim(0, 50)
ax.set_aspect('equal')

ax.set_xticks(np.arange(0, 21, 1))
ax.set_yticks(np.arange(0, 51, 1))
ax.grid(True, linestyle='--', linewidth=0.5)

# Draw original lines
for label, (p1, p2) in lines.items():
    ax.plot([p1[0], p2[0]], [p1[1], p2[1]])
    mx, my = (p1[0]+p2[0])/2, (p1[1]+p2[1])/2
    ax.text(mx, my, label, fontsize=9)

# ----------------------------
# Alpha shape (concave hull)
# ----------------------------
def alpha_shape(points, alpha):
    if len(points) < 4:
        return points

    tri = Delaunay(points)
    edges = set()

    def edge(p1, p2):
        if (p1, p2) in edges or (p2, p1) in edges:
            return
        edges.add((p1, p2))

    for ia, ib, ic in tri.simplices:
        pa, pb, pc = points[ia], points[ib], points[ic]

        a = np.linalg.norm(pb - pa)
        b = np.linalg.norm(pc - pb)
        c = np.linalg.norm(pa - pc)

        s = (a + b + c) / 2.0
        area = max(s * (s-a) * (s-b) * (s-c), 1e-12)
        circum_r = (a * b * c) / (4.0 * math.sqrt(area))

        # Keep triangles that are "tight enough"
        if circum_r < 1.0 / alpha:
            edge(tuple(pa), tuple(pb))
            edge(tuple(pb), tuple(pc))
            edge(tuple(pc), tuple(pa))

    # Build boundary (simple ordering by angle)
    boundary = list(edges)
    pts = np.array(list(set([p for e in boundary for p in e])))

    center = pts.mean(axis=0)
    angles = np.arctan2(pts[:,1] - center[1], pts[:,0] - center[0])
    sorted_pts = pts[np.argsort(angles)]

    return sorted_pts

# Tune this:
alpha = 0.08

hull = alpha_shape(points, alpha)

# close loop
hx = np.append(hull[:,0], hull[0,0])
hy = np.append(hull[:,1], hull[0,1])

ax.plot(hx, hy, color='red', linewidth=2)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
from scipy.spatial import Delaunay
import math

# -------------------------
# DATA
# -------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

weights = {
    "pd1": 0,
    "pd2": 30,
    "pd3": 60,
    "pd4": 80,
    "pd5": 40,
    "pd6": 10,
    "pd7": 30,
    "pd8": 0,
}

# -------------------------
# GRID
# -------------------------
W, H = 300, 750

x = np.linspace(0, 20, W)
y = np.linspace(0, 50, H)

X, Y = np.meshgrid(x, y)

# -------------------------
# CONCAVE HULL (alpha shape simplified)
# -------------------------
def concave_hull(points, alpha=0.12):
    pts = np.array(points)
    tri = Delaunay(pts)

    edges = set()

    def add_edge(a, b):
        if (a, b) not in edges and (b, a) not in edges:
            edges.add((a, b))

    for ia, ib, ic in tri.simplices:
        pa, pb, pc = pts[ia], pts[ib], pts[ic]

        a = np.linalg.norm(pb - pa)
        b = np.linalg.norm(pc - pb)
        c = np.linalg.norm(pa - pc)

        s = (a + b + c) / 2
        area = max(s * (s-a) * (s-b) * (s-c), 1e-12)
        R = (a*b*c) / (4 * math.sqrt(area))

        if R < 1 / alpha:
            add_edge(tuple(pa), tuple(pb))
            add_edge(tuple(pb), tuple(pc))
            add_edge(tuple(pc), tuple(pa))

    pts2 = np.array(list(set([p for e in edges for p in e])))

    center = pts2.mean(axis=0)
    angles = np.arctan2(pts2[:,1] - center[1], pts2[:,0] - center[0])

    return pts2[np.argsort(angles)]

# endpoints → hull
all_points = [p for seg in lines.values() for p in seg]
hull = concave_hull(all_points, alpha=0.12)

# -------------------------
# MASK FROM HULL (point-in-polygon)
# -------------------------
def point_in_poly(x, y, poly):
    inside = False
    n = len(poly)

    for i in range(n):
        x1,y1 = poly[i]
        x2,y2 = poly[(i+1)%n]

        if ((y1 > y) != (y2 > y)) and \
           (x < (x2-x1)*(y-y1)/(y2-y1+1e-9) + x1):
            inside = not inside

    return inside

mask = np.zeros((H, W))

for i in range(H):
    for j in range(W):
        if point_in_poly(X[i,j], Y[i,j], hull):
            mask[i,j] = 1

# soften boundary (IMPORTANT for realism)
mask = gaussian_filter(mask, sigma=4)

# -------------------------
# PRESSURE FIELD (your smooth model)
# -------------------------
def build_profile(line_map, axis, sigma=3.0):
    profile = np.zeros_like(axis)

    for name, pos in line_map.items():
        profile += weights[name] * np.exp(-0.5 * ((axis - pos) / sigma)**2)

    profile = gaussian_filter(profile, sigma=5)
    return profile / (profile.max() + 1e-9)

vertical_lines = {
    "pd1": 15,
    "pd2": 11,
    "pd3": 7,
}

horizontal_lines = {
    "pd4": 6,
    "pd5": 14,
    "pd6": 22,
    "pd7": 30,
    "pd8": 38,
}

V = build_profile(vertical_lines, x)
Hh = build_profile(horizontal_lines, y)

pressure = np.outer(Hh, V)

# extra diffusion for tissue-like feel
pressure = gaussian_filter(pressure, sigma=5)

# -------------------------
# APPLY FOOT MASK (keep NaNs instead of zeros)
# -------------------------
pressure_masked = np.where(mask > 0.1, pressure, np.nan)

# slight gamma for realism (only inside valid region)
pressure_masked = np.where(np.isnan(pressure_masked), np.nan, pressure_masked ** 1.2)

# -------------------------
# PLOT
# -------------------------
cmap = plt.cm.hot.copy()
cmap.set_bad(color="white")  # <-- outside hull color (change as you like)

plt.figure(figsize=(6, 12))
plt.imshow(
    pressure_masked,
    origin="lower",
    extent=[0, 20, 0, 50],
    cmap=cmap,
    aspect="auto"
)

plt.colorbar(label="Pressure")
plt.title("Foot Pressure Map with Concave Hull Constraint")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter
import math

# -------------------------
# DATA (UPDATED)
# -------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

weights = {
    "pd1": 0,
    "pd2": 30,
    "pd3": 60,
    "pd4": 80,
    "pd5": 40,
    "pd6": 10,
    "pd7": 0,
    "pd8": 0,
}

# -------------------------
# GRID (coarse = "real heatmap cells")
# -------------------------
W, H = 80, 180   # << key change (cell-based resolution)

x = np.linspace(0, 20, W)
y = np.linspace(0, 50, H)

X, Y = np.meshgrid(x, y)

# -------------------------
# SIMPLE HULL (skip expensive triangulation for grid model)
# -------------------------
def point_in_poly(x, y, poly):
    inside = False
    n = len(poly)

    for i in range(n):
        x1,y1 = poly[i]
        x2,y2 = poly[(i+1)%n]

        if ((y1 > y) != (y2 > y)) and \
           (x < (x2-x1)*(y-y1)/(y2-y1+1e-9) + x1):
            inside = not inside

    return inside

# -------------------------
# FOOT PROFILE MODEL
# -------------------------
def build_profile(axis, lines_map, sigma=2.5):
    profile = np.zeros_like(axis)

    for name, pos in lines_map.items():
        profile += weights[name] * np.exp(-0.5 * ((axis - pos) / sigma)**2)

    profile = gaussian_filter(profile, sigma=2)
    return profile / (profile.max() + 1e-9)

vertical_lines = {
    "pd1": 15,
    "pd2": 11,
    "pd3": 6,
}

horizontal_lines = {
    "pd4": 6,
    "pd5": 14,
    "pd6": 24,
    "pd7": 30,
    "pd8": 38,
}

V = build_profile(x, vertical_lines)
Hh = build_profile(y, horizontal_lines)

# -------------------------
# GRID PRESSURE (cell-based)
# -------------------------
pressure = np.outer(Hh, V)

# soften like tissue
pressure = gaussian_filter(pressure, sigma=1.2)

# -------------------------
# FOOT MASK (define hull in grid space)
# -------------------------
foot_outline = [
    (15, 1), (16, 44), (19, 38), (19, 30),
    (18, 24), (18, 14), (18, 6),
    (16, 44), (12, 49), (9, 47), (4, 47),
    (1, 38), (2, 30), (6, 24), (7, 14),
    (6, 6), (15, 1)
]

mask = np.zeros((H, W))

for i in range(H):
    for j in range(W):
        if point_in_poly(X[i, j], Y[i, j], foot_outline):
            mask[i, j] = 1

# soften edge slightly (keeps grid look but avoids jaggies)
mask = gaussian_filter(mask, sigma=1.0)
mask = (mask > 0.3).astype(float)

# -------------------------
# APPLY MASK WITH DISTINCT NULL VALUE
# -------------------------
pressure = np.where(mask > 0, pressure, np.nan)

# gamma inside only
pressure = np.where(np.isnan(pressure), np.nan, pressure ** 1.15)

# -------------------------
# UPSAMPLE FOR VISUAL SMOOTHNESS (optional but key look)
# -------------------------
pressure_vis = gaussian_filter(np.nan_to_num(pressure, nan=0.0), sigma=1.2)
pressure_vis[mask == 0] = np.nan

# -------------------------
# PLOT
# -------------------------
cmap = plt.cm.hot.copy()
cmap.set_bad(color="#111111")  # clear null region

plt.figure(figsize=(6, 12))
plt.imshow(
    pressure_vis,
    origin="lower",
    extent=[0, 20, 0, 50],
    cmap=cmap,
    aspect="auto",
    interpolation="nearest"  # IMPORTANT: keeps grid feel
)

plt.colorbar(label="Pressure")
plt.title("Grid-Based Foot Pressure Heatmap")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# -------------------------
# DATA
# -------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

weights = {
    "pd1": 60,
    "pd2": 30,
    "pd3": 60,
    "pd4": 80,
    "pd5": 40,
    "pd6": 10,
    "pd7": 60,
    "pd8": 60,
}

# -------------------------
# GRID
# -------------------------
W, H = 90, 200

x = np.linspace(0, 20, W)
y = np.linspace(0, 50, H)
X, Y = np.meshgrid(x, y)

# -------------------------
# PRESSURE FIELD
# -------------------------
def build_profile(axis, lines_map, sigma=2.5):
    profile = np.zeros_like(axis)

    for name, pos in lines_map.items():
        profile += weights[name] * np.exp(-0.5 * ((axis - pos) / sigma)**2)

    profile = gaussian_filter(profile, sigma=2)
    return profile / (profile.max() + 1e-9)

vertical_lines = {"pd1": 15, "pd2": 11, "pd3": 6}
horizontal_lines = {
    "pd4": 6,
    "pd5": 14,
    "pd6": 24,
    "pd7": 30,
    "pd8": 38,
}

V = build_profile(x, vertical_lines)
Hh = build_profile(y, horizontal_lines)

pressure = np.outer(Hh, V)
pressure = gaussian_filter(pressure, sigma=1.2)

# -------------------------
# ENVELOPE-BASED FOOT MASK (FIX)
# -------------------------

# Build left/right boundaries from horizontal bands
bands = []
for name, yval in horizontal_lines.items():
    left = min(lines[name][0][0], lines[name][1][0])
    right = max(lines[name][0][0], lines[name][1][0])
    bands.append((yval, left, right))

bands = sorted(bands)

ys = np.array([b[0] for b in bands])
lefts = np.array([b[1] for b in bands])
rights = np.array([b[2] for b in bands])

# interpolate smooth envelope
left_env = np.interp(y, ys, lefts)
right_env = np.interp(y, ys, rights)

mask = np.zeros((H, W))

for i in range(H):
    for j in range(W):
        if left_env[i] <= X[i, j] <= right_env[i]:
            mask[i, j] = 1

mask = gaussian_filter(mask, sigma=1.0)
mask = (mask > 0.3).astype(float)

# -------------------------
# APPLY MASK
# -------------------------
pressure = np.where(mask > 0, pressure, np.nan)
pressure = np.where(np.isnan(pressure), np.nan, pressure ** 1.15)

# -------------------------
# VISUAL SMOOTHING ONLY
# -------------------------
pressure_vis = gaussian_filter(np.nan_to_num(pressure, nan=0.0), sigma=1.2)
pressure_vis[mask == 0] = np.nan

# -------------------------
# PLOT
# -------------------------
cmap = plt.cm.hot.copy()
cmap.set_bad(color="#121212")

plt.figure(figsize=(6, 12))
plt.imshow(
    pressure_vis,
    origin="lower",
    extent=[0, 20, 0, 50],
    cmap=cmap,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Pressure")
plt.title("Foot Pressure Heatmap (Envelope-Based Stable Mask)")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# -------------------------
# DATA
# -------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

weights = {
    "pd1": 20,
    "pd2": 60,
    "pd3": 80,
    "pd4": 20,
    "pd5": 20,
    "pd6": 20,
    "pd7": 80,
    "pd8": 80,
}

# -------------------------
# LOWER-RES GRID (KEY CHANGE)
# -------------------------
W, H = 45, 100   # << fewer cells = more "sensor plate" look

x = np.linspace(0, 20, W)
y = np.linspace(0, 50, H)

X, Y = np.meshgrid(x, y)

# -------------------------
# PRESSURE MODEL (VERTICAL-HEAVY)
# -------------------------
def build_profile(axis, lines_map, sigma=1.6, gain=1.8):
    profile = np.zeros_like(axis)

    for name, pos in lines_map.items():
        w = weights[name]

        # emphasize vertical structure
        if name in ["pd1", "pd2", "pd3"]:
            w *= gain

        profile += w * np.exp(-0.5 * ((axis - pos) / sigma)**2)

    profile = gaussian_filter(profile, sigma=1.5)
    return profile / (profile.max() + 1e-9)

vertical_lines = {"pd1": 15, "pd2": 11, "pd3": 6}

horizontal_lines = {
    "pd4": 6,
    "pd5": 14,
    "pd6": 24,
    "pd7": 30,
    "pd8": 38,
}

V = build_profile(x, vertical_lines)
Hh = build_profile(y, horizontal_lines)

# -------------------------
# BASE PRESSURE FIELD
# -------------------------
pressure = np.outer(Hh, V)

# -------------------------
# ANISOTROPIC ELONGATION (IMPORTANT)
# -------------------------
pressure = gaussian_filter(pressure, sigma=(8, 4))
#                           (x-tight, y-stretch)

# -------------------------
# ENVELOPE MASK (NO HULL)
# -------------------------
bands = []
for name, yval in horizontal_lines.items():
    left = min(lines[name][0][0], lines[name][1][0])
    right = max(lines[name][0][0], lines[name][1][0])
    bands.append((yval, left, right))

bands = sorted(bands)

ys = np.array([b[0] for b in bands])
lefts = np.array([b[1] for b in bands])
rights = np.array([b[2] for b in bands])

left_env = np.interp(y, ys, lefts)
right_env = np.interp(y, ys, rights)

mask = np.zeros((H, W))

for i in range(H):
    for j in range(W):
        if left_env[i] <= X[i, j] <= right_env[i]:
            mask[i, j] = 1

mask = gaussian_filter(mask, sigma=0.8)
mask = (mask > 0.25).astype(float)

# -------------------------
# APPLY MASK
# -------------------------
pressure = np.where(mask > 0, pressure, np.nan)
pressure = np.where(np.isnan(pressure), np.nan, pressure ** 1.12)

# -------------------------
# VISUAL SMOOTHING ONLY
# -------------------------
pressure_vis = gaussian_filter(np.nan_to_num(pressure, nan=0.0), sigma=1.0)
pressure_vis[mask == 0] = np.nan

# -------------------------
# PLOT
# -------------------------
cmap = plt.cm.hot.copy()
cmap.set_bad(color="white")

plt.figure(figsize=(6, 12))
plt.imshow(
    pressure_vis,
    origin="lower",
    extent=[0, 20, 0, 50],
    cmap=cmap,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Pressure")
plt.title("Low-Resolution Elongated Foot Pressure Heatmap")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.ndimage import gaussian_filter

# -------------------------
# DATA
# -------------------------
lines = {
    "pd1": ((15, 1), (16, 44)),
    "pd2": ((12, 0), (10, 49)),
    "pd3": ((9, 1), (4, 47)),
    "pd4": ((6, 6), (18, 6)),
    "pd5": ((7, 14), (18, 14)),
    "pd6": ((6, 24), (18, 24)),
    "pd7": ((2, 30), (19, 30)),
    "pd8": ((1, 38), (19, 38)),
}

weights = {
    "pd1": 0,
    "pd2": 30,
    "pd3": 60,
    "pd4": 80,
    "pd5": 40,
    "pd6": 10,
    "pd7": 0,
    "pd8": 0,
}

# -------------------------
# GRID (simple heatmap)
# -------------------------
W, H = 30, 70

x = np.linspace(0, 20, W)
y = np.linspace(0, 50, H)

X, Y = np.meshgrid(x, y)

# -------------------------
# SIMPLE PRESSURE FIELD
# -------------------------
def build_profile(axis, lines_map, sigma=1.8):
    profile = np.zeros_like(axis)

    for name, pos in lines_map.items():
        profile += weights[name] * np.exp(-0.5 * ((axis - pos) / sigma)**2)

    return gaussian_filter(profile, sigma=1.5)

vertical_lines = {"pd1": 15, "pd2": 11, "pd3": 6}

horizontal_lines = {
    "pd4": 6,
    "pd5": 14,
    "pd6": 24,
    "pd7": 30,
    "pd8": 38,
}

V = build_profile(x, vertical_lines)
Hh = build_profile(y, horizontal_lines)

pressure = np.outer(Hh, V)
pressure = gaussian_filter(pressure, sigma=(8, 4))



# -------------------------
# BUILD SIMPLE OUTER BOUNDARY
# -------------------------
pts = np.array([p for seg in lines.values() for p in seg])

# sort by angle around centroid (simple convex-ish boundary, but stable)
center = pts.mean(axis=0)
angles = np.arctan2(pts[:,1] - center[1], pts[:,0] - center[0])
poly = pts[np.argsort(angles)]

# -------------------------
# POINT IN POLYGON
# -------------------------
def point_in_poly(x, y, poly):
    inside = False
    n = len(poly)

    for i in range(n):
        x1, y1 = poly[i]
        x2, y2 = poly[(i + 1) % n]

        if ((y1 > y) != (y2 > y)) and \
           (x < (x2 - x1) * (y - y1) / (y2 - y1 + 1e-9) + x1):
            inside = not inside

    return inside

# -------------------------
# MASK (WHITE OUTSIDE)
# -------------------------
mask = np.zeros((H, W))

for i in range(H):
    for j in range(W):
        if point_in_poly(X[i, j], Y[i, j], poly):
            mask[i, j] = 1

mask = gaussian_filter(mask, sigma=0.6)
mask = (mask > 0.3).astype(float)

# -------------------------
# APPLY MASK
# -------------------------
pressure = np.where(mask > 0, pressure, np.nan)

# -------------------------
# VISUAL
# -------------------------
pressure_vis = gaussian_filter(np.nan_to_num(pressure, nan=0.0), sigma=1.0)
pressure_vis[mask == 0] = np.nan

cmap = plt.cm.hot.copy()
cmap.set_bad(color="white")

plt.figure(figsize=(6, 12))
plt.imshow(
    pressure_vis,
    origin="lower",
    extent=[0, 20, 0, 50],
    cmap=cmap,
    aspect="auto",
    interpolation="nearest"
)

plt.colorbar(label="Pressure")
plt.title("Simple Foot Mask (Outer Boundary Only)")
plt.show()
```
